#Crie um array de 10 elementos e extraia os 5 primeiros elementos e os 5 últimos elementos

```python
import numpy as np
array=np.arange(0,10)
print(array)
```

#Extraia os 5 primeiros elemento

```python
primeiros_elementos=array[:5]
print(primeiros_elementos)
```

#Extraia os 5 últimos elemento

```python
ultimos_elementos= array[-5:]
print(ultimos_elementos)
```

 #Crie um array com 10 números aleatórios e calcule o valor máximo, mínimo e a mediana.

```python
array_aleatorio=np.random.randint(10,100,10)
print("Array aleatório:")
print(array_aleatorio)
```

#Calcula o valor máximo

```python
valor_maximo=np.max(array_aleatorio)
print("Valor máximo:")
print(valor_maximo)
```

#Calcule o valor mínimo

```python
valor_mínimo=np.min(array_aleatorio)
print("Vaor mínimo:")
print(valor_mínimo)
```

#Calcule a mediana

```python
mediana= np.median(array_aleatorio)
print("Mediana:")
print(mediana)
```
