**Criar um array unidimensional usando range com 20 elementos**

```python
array=np.arange(20)
print(array)
```

**“Varrer” o array para obter os pares**

```python
import numpy as np

pares = []
array = np.arange(20)

for numero in array:
    if numero % 2 == 0:
        pares.append(numero)

print("Números pares:", pares)
```

**“Varrer” o array para obter os ímpares**

```python
impares=[]
for numero in array:
  if numero % 2 !=0:
    impares.append(numero)
    print("Numeros ímpares:",impares)
```

**Obter a média, a soma, o produto, a variância e o desvio padrão do array**

```python
media=np.mean(array)
soma=np.sum(array)
produto=np.prod(array)
variancia=np.var(array)
desvio_padrao=np.std(array)

print(f"Média:{media}")
print(f"Soma:{soma}")
print(f"Produto:{produto}")
print(f"Variância:{variancia}")
print(f"Desvio Padrão:{desvio_padrao}")
```
